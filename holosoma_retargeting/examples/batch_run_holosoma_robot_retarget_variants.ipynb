# Batch Run Robot Retarget Scripts

Use this notebook to run several `robot_retarget*.py` scripts with the same task settings.

Edit `SCRIPTS` to choose which files to run, then run the last cell.

```python
from pathlib import Path
import subprocess
import time

ROOT = Path("/home/cipher/Codes/Academic-Projects/Retarget-whole")
HOLOSOMA = ROOT / "holosoma_retargeting" / "holosoma_retargeting"
EXAMPLES = HOLOSOMA / "examples"

PYTHON = ["conda", "run", "--live-stream", "-n", "robot", "python"]
```

## Settings

```python
TASK_TYPE = "object_interaction"
ROBOT = "g1"
TASK_NAME = "sub3_largebox_003"
DATA_PATH = HOLOSOMA / "demo_data" / "OMOMO_new"

COMMON_ARGS = [
    "--task-type", TASK_TYPE,
    "--robot", ROBOT,
    "--task-name", TASK_NAME,
    "--data-path", str(DATA_PATH),
    "--retargeter.no-visualize",
    "--retargeter.no-debug",
]
```

## Scripts To Run

Delete or comment out any file you do not want to run.

```python
SCRIPTS = [
    "robot_retarget.py",
    "robot_retarget_etasp.py",
    "robot_retarget_first_order.py",
    "robot_retarget_first_order_etasp.py",
    "robot_retarget_first_order_no_trust_region.py",
    "robot_retarget_first_order_no_trust_region_etasp.py",
    "robot_retarget_laplacian_smooth.py",
    "robot_retarget_laplacian_smooth_etasp.py",
    "robot_retarget_laplacian_smooth_first_order.py",
    "robot_retarget_laplacian_smooth_first_order_etasp.py",
    "robot_retarget_laplacian_smooth_first_order_no_trust_region.py",
    "robot_retarget_laplacian_smooth_first_order_no_trust_region_etasp.py",
    "robot_retarget_laplacian_smooth_no_trust_region.py",
    "robot_retarget_laplacian_smooth_no_trust_region_etasp.py",
    "robot_retarget_no_trust_region.py",
    "robot_retarget_no_trust_region_etasp.py",
]
```

## Run

```python
successes = []
failures = []

for script_name in SCRIPTS:
    script_path = EXAMPLES / script_name
    if not script_path.exists():
        print(f"\nMissing {script_name}: {script_path}")
        failures.append((script_name, "missing", None))
        continue

    cmd = [*PYTHON, str(script_path), *COMMON_ARGS]

    print(f"\nRunning {script_name}")
    start = time.time()
    result = subprocess.run(
        cmd,
        cwd=HOLOSOMA,
        text=True,
    )
    elapsed = time.time() - start

    print(f"return code: {result.returncode}, time: {elapsed:.1f}s")

    if result.returncode != 0:
        failures.append((script_name, result.returncode, elapsed))
        print(f"Failed: {script_name}; continuing.")
        continue

    successes.append((script_name, elapsed))

print("\nDone.")
print(f"Succeeded: {len(successes)}")
for script_name, elapsed in successes:
    print(f"  OK     {script_name} ({elapsed:.1f}s)")

print(f"Failed: {len(failures)}")
for script_name, status, elapsed in failures:
    if elapsed is None:
        print(f"  FAIL   {script_name} ({status})")
    else:
        print(f"  FAIL   {script_name} (return code {status}, {elapsed:.1f}s)")
```
